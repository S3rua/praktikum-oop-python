```python
class Hero:
  # Constructor: Dijalankan saat Hero baru dibuat
  def __init__(self, name, hp, attack_power):
    self.name = name  # Nama Hero
    self.hp = hp  # Nyawa (Health Point)
    self.attack_power = attack_power  # Kekuatan Serangan

  # Method untuk menampilkan info hero
  def info(self):
    print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power} ")

  # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
  def serang(self, lawan):
    print(f"{self.name} menyerang {lawan.name}!")
    lawan.diserang(self.attack_power)

  # Method diserang: Menerima damage
  def diserang(self, damage):
    self.hp -= damage
    print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# -- Main Program --
# Membuat Object (Instansiasi)
hero1 = Hero("Layla", 500, 15)
hero2 = Hero("Zilong", 120, 20)

# Memanggil Method
hero1.info()
hero2.info()

# Tambah kode Output di akhir program
print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2) # Layla menyerang Zilong
hero2.serang(hero1) # Zilong membalas
```

```python
class Mage(Hero):
  def __init__(self, name, hp, attack_power, mana):
    super().__init__(name, hp, attack_power)
    self.mana = mana

  def info(self):
    print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

  def skill_fireball(self, lawan):
    if self.mana >= 20:
      print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
      self.mana -= 20
      lawan.diserang(self.attack_power * 2)
    else:
      print(f"{self.name} gagal skill! Mana tidak cukup.")

# -- Main Program Baru --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond)
eudora.skill_fireball(balmond)
```

• Jelaskan peran fungsi super() dalam menghubungkan data dari class Anak ke
class Induk!



Jika `super().__init__(name, hp, attack_power)` dihapus, akan muncul error `AttributeError: 'Mage' object has no attribute 'name'` saat `eudora.info()` dijalankan. Hal ini karena atribut `name`, `hp`, dan `attack_power` dibuat oleh constructor `Hero`, tetapi constructor tersebut tidak dipanggil.

`super()` berfungsi untuk memanggil constructor atau method dari class induk. Pada `Mage`, `super().__init__(...)` digunakan agar atribut dari `Hero` tetap dibuat, kemudian `Mage` dapat menambahkan atribut khusus seperti `mana`.






```python
class Hero:
  def __init__(self, nama, hp_awal):
    self.nama = nama
    # Enkapsulasi: HP bersifat Private (hanya bisa diakses di dalam class ini)
    self.__hp = hp_awal

  # GETTER: Cara resmi melihat HP
  def get_hp(self):
    return self.__hp

  # SETTER: Cara resmi mengubah HP (dengan validasi)
  def set_hp(self, nilai_baru):
    if nilai_baru < 0:
      self.__hp = 0  # HP tidak boleh negatif
    elif nilai_baru > 1000:
      print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
      self.__hp = 1000
    else:
      self.__hp = nilai_baru

  def diserang(self, damage):
    # Kita pakai setter/getter bahkan di dalam class sendiri agar aman
    sisa_hp = self.get_hp() - damage
    self.set_hp(sisa_hp)
    print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")

# -- Uji Coba --
hero1 = Hero("Layla", 100)
# hero1.__hp = 9999 # GAGAL (Tidak akan mengubah hp asli)
# print(hero1.__hp) # ERROR (Tidak bisa dibaca langsung)
hero1.set_hp(-50) # Coba set negatif
print(hero1.get_hp()) # Output: 0 (Karena dicegat oleh logika Setter)
```

```python
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
# Ini adalah KONTRAK. Semua turunan wajib punya method di bawah ini.
class GameUnit(ABC):

    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass


# 2. Implementasi pada Class Konkret
class Hero(GameUnit):

    def __init__(self, nama):
        self.nama = nama

    # Kita WAJIB membuat method serang
    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")


class Monster(GameUnit):

    def __init__(self, jenis):
        self.jenis = jenis

    # Implementasi serang versi Monster
    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")


# -- Uji Coba --

# unit = GameUnit()  # ERROR! Abstract class tidak bisa jadi objek.

h = Hero("Alucard")
m = Monster("Serigala")

h.info()
m.info()

h.serang("Monster")
m.serang("Hero")
```

```python
# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")


# Child Class 1
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")


# Child Class 2
class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")


# Child Class 3
class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")


# -- Penerapan Polymorphism --

# Kita punya daftar hero campuran
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord")
]

print("--- PERANG DIMULAI ---")

# Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
for pahlawan in pasukan:
    pahlawan.serang()
```

Jika serang() pada Archer diubah menjadi tembak_panah, maka saat looping:
pahlawan.serang()
akan muncul error:
AttributeError
Karena Archer sudah tidak memiliki method serang().
Nama method harus sama agar satu perintah yang sama dapat menghasilkan perilaku berbeda pada setiap child class.

```python
from abc import ABC, abstractmethod


# ==========================================
# 1. ABSTRACTION & ENKAPSULASI (Parent Class)
# ==========================================
class KamarHotel(ABC):

  def __init__(self, nama_kamar, stok_awal, harga_dasar):
    self.nama_kamar = nama_kamar
    # Encapsulation: Private Attributes
    self.__stok = 0
    self.__harga_dasar = harga_dasar

    # Menggunakan setter internal untuk validasi stok awal
    self.set_stok(stok_awal)

  # GETTER & SETTER UNTUK STOK
  def get_stok(self):
    return self.__stok

  def set_stok(self, jumlah):
    if jumlah < 0:
      print(
          f"Gagal update stok {self.nama_kamar}! Stok tidak boleh negatif"
          f" ({jumlah})."
      )
    else:
      self.__stok = jumlah
      print(f"Berhasil menambahkan stok {self.nama_kamar}: {jumlah} unit.")

  # GETTER HARGA DASAR
  def get_harga_dasar(self):
    return self.__harga_dasar

  # ABSTRACT METHODS (Kontrak Wajib)
  @abstractmethod
  def tampilkan_detail(self):
    pass

  @abstractmethod
  def hitung_harga_total(self, jumlah_malam):
    pass


# ==========================================
# 2. INHERITANCE & POLYMORPHISM (Child Class 1)
# ==========================================
class KamarDeluxe(KamarHotel):

  def __init__(self, nama_kamar, stok_awal, harga_dasar, fasilitas):
    super().__init__(nama_kamar, stok_awal, harga_dasar)
    self.fasilitas = fasilitas

  def hitung_pajak(self):
    # Pajak Deluxe: 10% dari harga dasar
    return self.get_harga_dasar() * 0.10

  def hitung_harga_total(self, jumlah_malam):
    harga_per_malam = self.get_harga_dasar() + self.hitung_pajak()
    return harga_per_malam * jumlah_malam

  def tampilkan_detail(self, jumlah_malam):
    pajak = self.hitung_pajak()
    subtotal = self.hitung_harga_total(jumlah_malam)
    print(
        f"[DELUXE] {self.nama_kamar} | Fasilitas: {self.fasilitas}\n"
        f"Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f} | Pajak (10%): Rp"
        f" {pajak:,.0f}\n"
        f"Menginap: {jumlah_malam} malam | Subtotal: Rp {subtotal:,.0f}".replace(
            ",", "."
        )
    )


# ==========================================
# 3. INHERITANCE & POLYMORPHISM (Child Class 2)
# ==========================================
class KamarStandard(KamarHotel):

  def __init__(self, nama_kamar, stok_awal, harga_dasar, kapasitas):
    super().__init__(nama_kamar, stok_awal, harga_dasar)
    self.kapasitas = kapasitas

  def hitung_pajak(self):
    # Pajak Standard: 5% dari harga dasar
    return self.get_harga_dasar() * 0.05

  def hitung_harga_total(self, jumlah_malam):
    harga_per_malam = self.get_harga_dasar() + self.hitung_pajak()
    return harga_per_malam * jumlah_malam

  def tampilkan_detail(self, jumlah_malam):
    pajak = self.hitung_pajak()
    subtotal = self.hitung_harga_total(jumlah_malam)
    print(
        f"[STANDARD] {self.nama_kamar} | Kapasitas: {self.kapasitas}\n"
        f"Harga Dasar/Malam: Rp {self.get_harga_dasar():,.0f} | Pajak (5%): Rp"
        f" {pajak:,.0f}\n"
        f"Menginap: {jumlah_malam} malam | Subtotal: Rp {subtotal:,.0f}".replace(
            ",", "."
        )
    )


# ==========================================
# 4. FUNGSI PEMESANAN (Polymorphic Helper)
# ==========================================
def proses_transaksi(daftar_pesanan):
  print("\nSTRUK PEMESANAN")
  total_tagihan = 0

  for i, item in enumerate(daftar_pesanan, start=1):
    kamar = item["kamar"]
    malam = item["malam"]

    print(f"{i}. ", end="")
    kamar.tampilkan_detail(malam)

    total_tagihan += kamar.hitung_harga_total(malam)

  print("-" * 50)
  formatted_total = f"Rp {total_tagihan:,.0f}".replace(",", ".")
  print(f"TOTAL TAGIHAN: {formatted_total}")


# ==========================================
# 5. ALUR PROGRAM (MAIN PROGRAM)
# ==========================================
if __name__ == "__main__":
  print("SETUP DATA KAMAR")

  # a) Admin membuat data kamar Deluxe & Standard
  kamar1 = KamarDeluxe("Kamar Deluxe Sea View", 10, 1500000, "Private Pool")

  # b) Admin mencoba mengisi stok negatif pada Kamar Standard
  kamar2 = KamarStandard("Kamar Standard Superior", -5, 500000, "2 Orang")

  # Koreksi stok Kamar Standard ke nilai valid
  kamar2.set_stok(20)

  # c) & d) Tamu memesan kamar (Deluxe 2 malam, Standard 1 malam)
  pesanan_tamu = [
      {"kamar": kamar1, "malam": 2},
      {"kamar": kamar2, "malam": 1},
  ]

  # Menjalankan pemrosesan transaksi dan mencetak output
  proses_transaksi(pesanan_tamu)
```
